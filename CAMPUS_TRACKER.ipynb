```python
import pandas as pd
import numpy as np

# --- GOOGLE DRIVE CONNECTION ---
# This links the Colab to the Google Drive so the data saves permanently
from google.colab import drive
drive.mount('/content/drive')

# This tell the program where to save the files in the Drive
tasks_file = '/content/drive/MyDrive/tasks.txt'
events_file = '/content/drive/MyDrive/events.txt'
# ---------------------------------

# ============================================================
# Campus & Project Tracker [ CampusTracker ]
# ============================================================

# 1. LISTS: Empty lists is used to store dictionaries of our data
tasks = []
events = []

# 2. FILE & EXCEPTION HANDLING: Try to load existing data
try:
    f = open(tasks_file, "r")
    for line in f:
        line = line.strip()
        if line:
            p = line.split("|")
            # 3. DICTIONARY: Store each task with its details
            d = {"subject": p[0], "name": p[1], "hours": float(p[2]), "status": p[3]}
            tasks.append(d)
    f.close()
except FileNotFoundError:
    print("No old task data found. Starting fresh.")

try:
    f = open(events_file, "r")
    for line in f:
        line = line.strip()
        if line:
            p = line.split("|")
            d = {"type": p[0], "title": p[1], "date": p[2]}
            events.append(d)
    f.close()
except FileNotFoundError:
    print("No old event data found.")

# Function to save the data back to the files
def save_data():
    try:
        f = open(tasks_file, "w")
        for t in tasks:
            f.write(t["subject"] + "|" + t["name"] + "|" + str(t["hours"]) + "|" + t["status"] + "\n")
        f.close()

        f = open(events_file, "w")
        for e in events:
            f.write(e["type"] + "|" + e["title"] + "|" + e["date"] + "\n")
        f.close()
        print("Data saved successfully to your Google Drive.")
    except Exception as e:
        print("Error saving data:", e)

# 4. LOOPS: The main menu loop
while True:
    print("\n--- Campus & Project Tracker ---")
    print("1. Add Academic Task")
    print("2. Add Hackathon or Event")
    print("3. Show All Tasks")
    print("4. Show All Events")
    print("5. Search (Strings)")
    print("6. Update Task Status")
    print("7. Delete an Entry")
    print("8. Show Unique Subjects (Sets)")
    print("9. Generate Workload Report (Pandas & NumPy)")
    print("10. Save and Exit")

    choice = input("Enter choice (1-10): ")

    if choice == "1":
        # ---- Code to add a Task ----
        sub = input("Subject (e.g., Python, Math): ")
        name = input("Task Name: ")
        try:
            hrs = float(input("Expected hours to complete: "))
        except ValueError:
            print("Not a valid number. Setting to 1.0 hour.")
            hrs = 1.0
        status = "Pending"
        tasks.append({"subject": sub, "name": name, "hours": hrs, "status": status})
        print("Task added!")

    elif choice == "2":
        # ---- Code to add an Event ----
        etype = input("Event Type (e.g., Hackathon, Workshop): ")
        title = input("Event Title: ")
        date = input("Date (DD-MM-YYYY): ")
        events.append({"type": etype, "title": title, "date": date})
        print("Event added!")

    elif choice == "3":
        # ---- Code to show Tasks ----
        if len(tasks) == 0:
            print("No tasks right now.")
        else:
            n = 1
            for t in tasks:
                print(n, "-", t["subject"], "|", t["name"], "|", t["hours"], "hrs |", t["status"])
                n += 1

    elif choice == "4":
        # ---- Code to show Events ----
        if len(events) == 0:
            print("No events scheduled.")
        else:
            n = 1
            for e in events:
                print(n, "-", e["type"], "|", e["title"], "|", e["date"])
                n += 1

    elif choice == "5":
        # 5. STRINGS: Search using lower() to ignore capitalization
        key = input("Enter search word: ").lower()
        print("\n-- Task Results --")
        for t in tasks:
            if key in t["name"].lower() or key in t["subject"].lower():
                print(t["name"], "(", t["subject"], ")")

        print("-- Event Results --")
        for e in events:
            if key in e["title"].lower() or key in e["type"].lower():
                print(e["title"], "on", e["date"])

    elif choice == "6":
        # ---- Update Task ----
        n = 1
        for t in tasks:
            print(n, "-", t["name"], "- Current Status:", t["status"])
            n += 1
        try:
            num = int(input("Which task number is completed? "))
            if 1 <= num <= len(tasks):
                tasks[num-1]["status"] = "Completed"
                print("Task marked as completed!")
            else:
                print("Invalid number.")
        except ValueError:
            print("Please enter a valid number.")

    elif choice == "7":
        # ---- Delete ----
        print("1. Delete Task  2. Delete Event")
        c = input("Choice: ")
        if c == "1":
            try:
                num = int(input("Enter task number to delete: "))
                if 1 <= num <= len(tasks):
                    tasks.pop(num-1)
                    print("Task deleted.")
            except ValueError:
                print("Invalid input.")
        elif c == "2":
            try:
                num = int(input("Enter event number to delete: "))
                if 1 <= num <= len(events):
                    events.pop(num-1)
                    print("Event deleted.")
            except ValueError:
                print("Invalid input.")

    elif choice == "8":
        # 6. SETS: Extract unique subjects to remove duplicates
        subject_list = []
        for t in tasks:
            subject_list.append(t["subject"])

        unique_subjects = set(subject_list)
        print("\nYou are currently studying these unique subjects:")
        for sub in unique_subjects:
            print("-", sub)

    elif choice == "9":
        # 7. PANDAS & NUMPY: Generate an analytical report
        save_data() # Force a save so Pandas reads the latest data
        try:
            print("\n--- Workload Analytics ---")
            # Pandas: Read the text file using '|' as the separator
            df = pd.read_csv(tasks_file, sep="|", names=["Subject", "Name", "Hours", "Status"])

            # Display a clean table using Pandas
            print("Dataframe Preview:")
            print(df)

            # NumPy: Calculate statistics on the Hours column
            hours_array = df["Hours"].to_numpy()
            total_workload = np.sum(hours_array)
            average_time = np.mean(hours_array)

            print("\nTotal estimated hours of work:", total_workload)
            print(f"Average time per task: {average_time:.2f} hours")

        except Exception as e:
            print("Could not generate report. Make sure you have added tasks.")
            print("Error:", e)

    elif choice == "10":
        # ---- Exit ----
        save_data()
        print("Exiting Tracker. Goodbye!")
        break

    else:
        print("Invalid choice, try again.")
```
